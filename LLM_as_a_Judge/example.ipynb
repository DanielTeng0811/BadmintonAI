1. 周天成使用「殺球」得分佔他總得分的百分比是多少？

```python
# 檢查 df 是否不為空
if len(df) > 0:
    # 篩選周天成的主動得分
    tchou_active_wins = df[(df['player'] == 'CHOU Tien Chen') & (df['player'] == df['getpoint_player'])]
    
    # 計算「殺球」得分次數 (主動得分)
    smash_wins = tchou_active_wins[tchou_active_wins['type'] == '殺球']
    smash_win_count = len(smash_wins)
    
    # 計算總主動得分次數
    total_active_win_count = len(tchou_active_wins)

    # 計算「殺球」得分佔總得分的百分比
    if total_active_win_count > 0:
        smash_percentage = (smash_win_count / total_active_win_count) * 100
    else:
        smash_percentage = 0
        
    # 打印結果
    print(f"周天成使用「殺球」得分佔他總得分的百分比為: {smash_percentage:.2f}%")
else:
    print("資料集為空，無法計算。")
```
