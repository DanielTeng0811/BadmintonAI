```python
import pandas as pd

# 確保資料不為空
if len(df) > 0:
    # 過濾出周天成接高遠球的情況（場地區域定義中後場）
    filtered_df = df[(df['player'] == 'CHOU Tien Chen') & (df['hit_area'].isin([1, 2, 3, 4]))]

    # 需要獲取到前一球擊球的落點位置
    filtered_df['previous_landing_area'] = filtered_df.groupby(['match_id', 'set', 'rally'])['landing_area'].shift(1)
    
    # 定義直線和斜線
    def shot_type(row):
        if row['previous_landing_area'] is not None:
            previous_col = (row['previous_landing_area'] - 1) % 4  # 確定前一球位置的列
            current_col = (row['hit_area'] - 1) % 4  # 確定目前球位置的列
            if previous_col == current_col:
                return '直線'
            else:
                return '斜線'
        else:
            return '未知'

    # 應用我們的判斷 
    filtered_df['shot_type'] = filtered_df.apply(shot_type, axis=1)

    # 計算比例
    shot_type_counts = filtered_df['shot_type'].value_counts(normalize=True)

    # 輸出比例訊息
    print("直線與斜線的比例:")
    print(shot_type_counts)

else:
    print("Data is empty.")
```

```python
# 確保數據長度大於0
if len(df) > 0:
    # 1. 過濾周天成接發球的數據以及所需的type
    filtered_df = df[(df['player'] == 'CHOU Tien Chen') & 
                     (df['type'].isin(['長球', '網前球', '平球']))]
    
    # 檢查過濾後的數據長度
    if len(filtered_df) > 0:
        # 2. 計算每種策略的使用次數及比例
        strategy_counts = filtered_df['type'].value_counts()
        strategy_proportions = strategy_counts / strategy_counts.sum()
        
        print("各種策略的選擇次數：\n", strategy_counts)
        print("各種策略的比例：\n", strategy_proportions)
        
        # 3.1 計算得分的情況
        active_wins_df = filtered_df[filtered_df['player'] == filtered_df['getpoint_player']]
        win_strategy_counts = active_wins_df['type'].value_counts()
        
        # 3.2 計算每種策略對總得分的貢獻比例
        win_strategy_proportions = win_strategy_counts / win_strategy_counts.sum()
        
        print("每種策略得分次數：\n", win_strategy_counts)
        print("每種策略的得分比例：\n", win_strategy_proportions)
        
        # 4. 視覺化展示比例
        fig, axs = plt.subplots(1, 2, figsize=(12, 6))
        
        # 選擇比例的圓餅圖
        axs[0].pie(strategy_proportions, labels=strategy_proportions.index, autopct='%1.1f%%', startangle=90)
        axs[0].set_title('周天成接發球策略選擇比例')
        
        # 得分比例的條形圖
        import seaborn as sns
        sns.barplot(x=win_strategy_proportions.index, y=win_strategy_proportions, ax=axs[1])
        axs[1].set_title('周天成接發球策略得分比例')
        axs[1].set_ylabel('得分比例')
        
        plt.tight_layout()  # 確保布局不重疊
else:
    print("數據集中沒有可用的數據")
```
